```python
# ============================================================
# ORIGIN AI
# ============================================================
# DATASET FORENSICS, DIGITAL ARCHAEOLOGY & DATA INTELLIGENCE
# ============================================================
#
# "Every dataset has a history. ORIGIN reconstructs the evidence."
#
# COMPLETE SINGLE-CELL PROJECT
#
# Core capabilities:
#   01. Synthetic data generation
#   02. Dataset profiling
#   03. Dataset DNA
#   04. Data quality analysis
#   05. Missing-value forensics
#   06. Exact duplicate detection
#   07. Near-duplicate detection
#   08. Statistical outlier detection
#   09. Isolation Forest
#   10. Robust Z-score analysis
#   11. PCA
#   12. KMeans behavioral clustering
#   13. DBSCAN density analysis
#   14. Feature dependency analysis
#   15. Temporal drift detection
#   16. Distribution shift detection
#   17. Source fingerprinting
#   18. Record forensic scoring
#   19. Evidence engine
#   20. Investigation case generation
#   21. Dataset health scoring
#   22. Advanced visual analytics
#   23. Executive command center
#   24. Automated forensic report
#   25. CSV / JSON export
#
# IMPORTANT:
# This is an analytical/demo system using synthetic data.
# Statistical anomalies do NOT automatically imply fraud,
# malicious behavior, incorrect provenance, or causality.
# ============================================================

import warnings
warnings.filterwarnings("ignore")

import os
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy import stats
from scipy.spatial.distance import cdist

from sklearn.preprocessing import StandardScaler, RobustScaler
from sklearn.impute import SimpleImputer
from sklearn.ensemble import IsolationForest
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
from sklearn.neighbors import NearestNeighbors

from IPython.display import display, Markdown

# ============================================================
# 0. GLOBAL CONFIGURATION
# ============================================================

np.random.seed(42)

PROJECT_NAME = "ORIGIN AI"
PROJECT_VERSION = "2.0"

N_ROWS = 6000
RANDOM_STATE = 42

EXPORT_DIR = "origin_ai_outputs"

os.makedirs(
    EXPORT_DIR,
    exist_ok=True
)

print("=" * 110)
print("ORIGIN AI")
print("DATASET FORENSICS & DIGITAL DATA ARCHAEOLOGY ENGINE")
print("=" * 110)
print("Version:", PROJECT_VERSION)
print("Mode: Synthetic Research / Demonstration")
print("=" * 110)

# ============================================================
# 1. SYNTHETIC DATA GENERATION
# ============================================================

print("\n[1/25] Generating synthetic dataset...")

timestamps = pd.date_range(
    start="2026-01-01",
    periods=N_ROWS,
    freq="4h"
)

data = pd.DataFrame({
    "record_id": np.arange(
        1,
        N_ROWS + 1
    ),
    "timestamp": timestamps
})

# ------------------------------------------------------------
# Hidden source structure
# ------------------------------------------------------------

data["source"] = np.random.choice(
    [
        "SOURCE_A",
        "SOURCE_B",
        "SOURCE_C"
    ],
    size=N_ROWS,
    p=[
        0.48,
        0.34,
        0.18
    ]
)

# ------------------------------------------------------------
# Base numerical variables
# ------------------------------------------------------------

data["age"] = np.clip(
    np.random.normal(
        38,
        11,
        N_ROWS
    ),
    18,
    80
)

data["income"] = np.random.lognormal(
    mean=10.5,
    sigma=0.45,
    size=N_ROWS
)

data["activity_score"] = np.clip(
    np.random.normal(
        65,
        15,
        N_ROWS
    ),
    0,
    100
)

data["risk_score"] = np.random.beta(
    3,
    7,
    N_ROWS
)

data["transaction_count"] = np.random.poisson(
    18,
    N_ROWS
)

data["session_duration"] = np.random.gamma(
    4,
    8,
    N_ROWS
)

data["engagement"] = np.clip(
    0.6 * data["activity_score"]
    +
    np.random.normal(
        0,
        10,
        N_ROWS
    ),
    0,
    100
)

# ------------------------------------------------------------
# Correlated feature system
# ------------------------------------------------------------

data["feature_01"] = (
    0.40 * data["activity_score"]
    +
    0.30 * data["engagement"]
    +
    np.random.normal(
        0,
        7,
        N_ROWS
    )
)

data["feature_02"] = (
    data["income"] / 10000
    +
    0.2 * data["age"]
    +
    np.random.normal(
        0,
        4,
        N_ROWS
    )
)

data["feature_03"] = (
    0.70 * data["age"]
    +
    np.random.normal(
        0,
        7,
        N_ROWS
    )
)

data["feature_04"] = (
    1.8 * data["transaction_count"]
    +
    0.4 * data["activity_score"]
    +
    np.random.normal(
        0,
        5,
        N_ROWS
    )
)

data["feature_05"] = (
    data["risk_score"] * 100
    +
    np.random.normal(
        0,
        5,
        N_ROWS
    )
)

data["feature_06"] = (
    data["session_duration"]
    +
    np.random.normal(
        0,
        5,
        N_ROWS
    )
)

data["feature_07"] = (
    data["feature_01"] * 0.7
    +
    np.random.normal(
        0,
        5,
        N_ROWS
    )
)

# ------------------------------------------------------------
# Categorical variables
# ------------------------------------------------------------

data["region"] = np.random.choice(
    [
        "North",
        "South",
        "East",
        "West"
    ],
    size=N_ROWS,
    p=[
        0.24,
        0.31,
        0.20,
        0.25
    ]
)

data["device"] = np.random.choice(
    [
        "Mobile",
        "Desktop",
        "Tablet"
    ],
    size=N_ROWS,
    p=[
        0.60,
        0.32,
        0.08
    ]
)

data["account_type"] = np.random.choice(
    [
        "Basic",
        "Premium",
        "Enterprise"
    ],
    size=N_ROWS,
    p=[
        0.55,
        0.35,
        0.10
    ]
)

data["status"] = np.random.choice(
    [
        "Active",
        "Inactive",
        "Pending"
    ],
    size=N_ROWS,
    p=[
        0.72,
        0.18,
        0.10
    ]
)

# ============================================================
# 2. HIDDEN DATA EVENTS
# ============================================================

print("[2/25] Injecting hidden data events...")

# ------------------------------------------------------------
# EVENT A — Temporal distribution shift
# ------------------------------------------------------------

event_a = (
    data["timestamp"] >= "2026-03-20"
) & (
    data["timestamp"] <= "2026-03-27"
)

data.loc[
    event_a,
    "activity_score"
] += np.random.uniform(
    15,
    30,
    event_a.sum()
)

data.loc[
    event_a,
    "transaction_count"
] += np.random.randint(
    8,
    20,
    event_a.sum()
)

data.loc[
    event_a,
    "risk_score"
] += np.random.uniform(
    0.08,
    0.20,
    event_a.sum()
)

# ------------------------------------------------------------
# EVENT B — Source profile change
# ------------------------------------------------------------

event_b = (
    data["timestamp"] >= "2026-04-10"
) & (
    data["timestamp"] <= "2026-04-17"
)

data.loc[
    event_b,
    "source"
] = np.random.choice(
    [
        "SOURCE_X",
        "SOURCE_Y"
    ],
    event_b.sum()
)

data.loc[
    event_b,
    "feature_01"
] *= 1.8

data.loc[
    event_b,
    "feature_02"
] *= 0.55

data.loc[
    event_b,
    "feature_07"
] += 20

# ------------------------------------------------------------
# EVENT C — Outlier injection
# ------------------------------------------------------------

outlier_indices = np.random.choice(
    data.index,
    size=65,
    replace=False
)

data.loc[
    outlier_indices,
    "income"
] *= np.random.uniform(
    4,
    12,
    len(outlier_indices)
)

data.loc[
    outlier_indices,
    "transaction_count"
] *= np.random.randint(
    4,
    8,
    len(outlier_indices)
)

data.loc[
    outlier_indices,
    "session_duration"
] *= np.random.uniform(
    3,
    7,
    len(outlier_indices)
)

# ------------------------------------------------------------
# EVENT D — Missingness pattern
# ------------------------------------------------------------

missing_indices = np.random.choice(
    data.index,
    size=320,
    replace=False
)

data.loc[
    missing_indices,
    "feature_03"
] = np.nan

data.loc[
    missing_indices[:170],
    "feature_04"
] = np.nan

data.loc[
    missing_indices[:80],
    "feature_07"
] = np.nan

# ------------------------------------------------------------
# EVENT E — Duplicated records
# ------------------------------------------------------------

duplicate_rows = data.sample(
    50,
    random_state=99
).copy()

duplicate_rows["record_id"] = np.arange(
    N_ROWS + 1,
    N_ROWS + 51
)

data = pd.concat(
    [
        data,
        duplicate_rows
    ],
    ignore_index=True
)

# ------------------------------------------------------------
# EVENT F — Rare categorical combination
# ------------------------------------------------------------

rare_indices = np.random.choice(
    data.index,
    size=40,
    replace=False
)

data.loc[
    rare_indices,
    "region"
] = "UNKNOWN_REGION"

data.loc[
    rare_indices,
    "device"
] = "UNKNOWN_DEVICE"

# ============================================================
# 3. BASIC DATASET INVENTORY
# ============================================================

print("[3/25] Building dataset inventory...")

numeric_cols = data.select_dtypes(
    include=np.number
).columns.tolist()

if "record_id" in numeric_cols:
    numeric_cols.remove(
        "record_id"
    )

categorical_cols = data.select_dtypes(
    exclude=np.number
).columns.tolist()

if "timestamp" in categorical_cols:
    categorical_cols.remove(
        "timestamp"
    )

dataset_rows = len(data)
dataset_columns = data.shape[1]

exact_duplicates = int(
    data.duplicated().sum()
)

missing_cells = int(
    data.isna().sum().sum()
)

total_cells = (
    data.shape[0]
    *
    data.shape[1]
)

missing_rate = (
    missing_cells
    /
    total_cells
)

print(
    f"Rows: {dataset_rows:,}"
)

print(
    f"Columns: {dataset_columns}"
)

print(
    f"Numeric features: {len(numeric_cols)}"
)

print(
    f"Categorical features: {len(categorical_cols)}"
)

print(
    f"Exact duplicates: {exact_duplicates:,}"
)

print(
    f"Missing rate: {missing_rate:.2%}"
)

# ============================================================
# 4. COLUMN FORENSIC PROFILE
# ============================================================

print("[4/25] Profiling columns...")

profile_rows = []

for column in data.columns:

    series = data[column]

    profile_rows.append({

        "column":
            column,

        "dtype":
            str(series.dtype),

        "unique_values":
            int(series.nunique()),

        "missing_count":
            int(series.isna().sum()),

        "missing_rate":
            float(series.isna().mean()),

        "duplicate_values":
            int(
                series.duplicated().sum()
            )
    })

profile_df = pd.DataFrame(
    profile_rows
)

# ============================================================
# 5. MISSING VALUE FORENSICS
# ============================================================

print("[5/25] Running missing-value forensics...")

missing_df = (
    data.isna()
    .sum()
    .sort_values(
        ascending=False
    )
    .reset_index()
)

missing_df.columns = [
    "feature",
    "missing_count"
]

missing_df["missing_rate"] = (
    missing_df["missing_count"]
    /
    len(data)
)

# ============================================================
# 6. EXACT DUPLICATE FORENSICS
# ============================================================

print("[6/25] Running duplicate forensics...")

duplicate_rows_df = (
    data[
        data.duplicated(
            keep=False
        )
    ]
    .sort_values(
        by=list(data.columns)
    )
)

# ============================================================
# 7. PREPARE NUMERIC MATRIX
# ============================================================

print("[7/25] Preparing analytical matrix...")

numeric_data = data[
    numeric_cols
].copy()

imputer = SimpleImputer(
    strategy="median"
)

numeric_imputed = imputer.fit_transform(
    numeric_data
)

standard_scaler = StandardScaler()

X_scaled = standard_scaler.fit_transform(
    numeric_imputed
)

robust_scaler = RobustScaler()

X_robust = robust_scaler.fit_transform(
    numeric_imputed
)

# ============================================================
# 8. ROBUST STATISTICAL OUTLIERS
# ============================================================

print("[8/25] Detecting statistical anomalies...")

median_values = np.median(
    numeric_imputed,
    axis=0
)

mad_values = np.median(
    np.abs(
        numeric_imputed
        -
        median_values
    ),
    axis=0
)

mad_values[
    mad_values == 0
] = 1e-9

robust_z = (
    0.6745
    *
    (
        numeric_imputed
        -
        median_values
    )
    /
    mad_values
)

robust_z_abs = np.abs(
    robust_z
)

data["max_robust_z"] = (
    robust_z_abs.max(
        axis=1
    )
)

data["statistical_outlier"] = (
    data["max_robust_z"] >= 3.5
)

# ============================================================
# 9. ISOLATION FOREST
# ============================================================

print("[9/25] Running Isolation Forest...")

isolation_forest = IsolationForest(
    n_estimators=350,
    contamination=0.035,
    max_samples="auto",
    random_state=RANDOM_STATE,
    n_jobs=-1
)

data["if_prediction"] = (
    isolation_forest.fit_predict(
        X_scaled
    )
)

data["if_raw_score"] = (
    -isolation_forest.score_samples(
        X_scaled
    )
)

data["if_anomaly"] = (
    data["if_prediction"] == -1
)

# Normalize Isolation Forest score

if_min = data[
    "if_raw_score"
].min()

if_max = data[
    "if_raw_score"
].max()

data["if_normalized_score"] = (
    data["if_raw_score"]
    -
    if_min
) / (
    if_max
    -
    if_min
    +
    1e-9
)

# ============================================================
# 10. PCA
# ============================================================

print("[10/25] Performing PCA...")

pca = PCA(
    n_components=2,
    random_state=RANDOM_STATE
)

pca_result = pca.fit_transform(
    X_scaled
)

data["PCA_1"] = (
    pca_result[:, 0]
)

data["PCA_2"] = (
    pca_result[:, 1]
)

pc1_variance = (
    pca.explained_variance_ratio_[0]
)

pc2_variance = (
    pca.explained_variance_ratio_[1]
)

# ============================================================
# 11. KMEANS BEHAVIORAL CLUSTERING
# ============================================================

print("[11/25] Discovering behavioral clusters...")

kmeans = KMeans(
    n_clusters=5,
    n_init=15,
    random_state=RANDOM_STATE
)

data["behavior_cluster"] = (
    kmeans.fit_predict(
        X_scaled
    )
)

# ============================================================
# 12. DBSCAN DENSITY ANALYSIS
# ============================================================

print("[12/25] Running density analysis...")

dbscan = DBSCAN(
    eps=1.65,
    min_samples=20
)

data["density_cluster"] = (
    dbscan.fit_predict(
        X_scaled
    )
)

data["density_outlier"] = (
    data["density_cluster"] == -1
)

# ============================================================
# 13. FEATURE DEPENDENCY ANALYSIS
# ============================================================

print("[13/25] Investigating feature dependencies...")

correlation = data[
    numeric_cols
].corr()

dependency_rows = []

for i in range(
    len(correlation.columns)
):

    for j in range(
        i + 1,
        len(correlation.columns)
    ):

        value = correlation.iloc[
            i,
            j
        ]

        if abs(value) >= 0.70:

            dependency_rows.append({

                "feature_1":
                    correlation.columns[i],

                "feature_2":
                    correlation.columns[j],

                "correlation":
                    float(value),

                "strength":
                    (
                        "Very Strong"
                        if abs(value) >= 0.85
                        else
                        "Strong"
                    )
            })

dependency_df = pd.DataFrame(
    dependency_rows
)

# ============================================================
# 14. TEMPORAL DISTRIBUTION ANALYSIS
# ============================================================

print("[14/25] Running temporal drift analysis...")

data["date"] = (
    data["timestamp"]
    .dt.date
)

daily_numeric = (
    data.groupby(
        "date"
    )[numeric_cols]
    .mean()
)

# Use first 7 days as baseline

baseline_days = min(
    7,
    len(daily_numeric)
)

baseline_profile = (
    daily_numeric
    .iloc[:baseline_days]
    .mean()
)

daily_drift_records = []

for date, row in daily_numeric.iterrows():

    feature_drifts = []

    for feature in numeric_cols:

        baseline_value = (
            baseline_profile[
                feature
            ]
        )

        current_value = (
            row[feature]
        )

        relative_change = (
            abs(
                current_value
                -
                baseline_value
            )
            /
            (
                abs(
                    baseline_value
                )
                +
                1e-9
            )
        )

        feature_drifts.append(
            relative_change
        )

    daily_drift_records.append({

        "date":
            date,

        "average_drift":
            np.mean(
                feature_drifts
            ),

        "maximum_drift":
            np.max(
                feature_drifts
            )
    })

daily_drift_df = pd.DataFrame(
    daily_drift_records
)

# ============================================================
# 15. DISTRIBUTION SHIFT TESTING
# ============================================================

print("[15/25] Running distribution-shift tests...")

distribution_shift_rows = []

baseline_data = data[
    data["timestamp"]
    <
    pd.Timestamp("2026-03-01")
]

current_data = data[
    data["timestamp"]
    >=
    pd.Timestamp("2026-03-01")
]

for feature in numeric_cols:

    baseline_values = (
        baseline_data[
            feature
        ]
        .dropna()
        .values
    )

    current_values = (
        current_data[
            feature
        ]
        .dropna()
        .values
    )

    if (
        len(baseline_values) >= 20
        and
        len(current_values) >= 20
    ):

        statistic, p_value = (
            stats.ks_2samp(
                baseline_values,
                current_values
            )
        )

        distribution_shift_rows.append({

            "feature":
                feature,

            "ks_statistic":
                float(statistic),

            "p_value":
                float(p_value),

            "shift_detected":
                bool(
                    p_value < 0.05
                )
        })

distribution_shift_df = pd.DataFrame(
    distribution_shift_rows
)

# ============================================================
# 16. SOURCE FINGERPRINTING
# ============================================================

print("[16/25] Building source fingerprints...")

source_profile = (
    data.groupby(
        "source"
    )[numeric_cols]
    .mean()
)

source_scaled = pd.DataFrame(
    StandardScaler().fit_transform(
        source_profile
    ),
    index=source_profile.index,
    columns=source_profile.columns
)

# Calculate pairwise source distance

source_distance_matrix = pd.DataFrame(
    cdist(
        source_scaled.values,
        source_scaled.values,
        metric="euclidean"
    ),
    index=source_scaled.index,
    columns=source_scaled.index
)

# ============================================================
# 17. RECORD-LEVEL FORENSIC SCORE
# ============================================================

print("[17/25] Calculating forensic scores...")

missing_per_row = (
    data[
        numeric_cols
    ]
    .isna()
    .sum(axis=1)
)

missing_score = (
    missing_per_row
    /
    len(numeric_cols)
)

statistical_score = np.clip(
    data[
        "max_robust_z"
    ]
    /
    10,
    0,
    1
)

density_score = (
    data[
        "density_outlier"
    ]
    .astype(float)
)

isolation_score = (
    data[
        "if_normalized_score"
    ]
)

data["forensic_score"] = (
    0.45 * isolation_score
    +
    0.30 * statistical_score
    +
    0.15 * missing_score
    +
    0.10 * density_score
)

data["forensic_score"] = (
    data[
        "forensic_score"
    ]
    .clip(0, 1)
)

# ============================================================
# 18. FORENSIC CLASSIFICATION
# ============================================================

def classify_risk(score):

    if score >= 0.80:
        return "CRITICAL"

    elif score >= 0.60:
        return "HIGH"

    elif score >= 0.35:
        return "MEDIUM"

    else:
        return "LOW"


data["forensic_level"] = (
    data[
        "forensic_score"
    ]
    .apply(
        classify_risk
    )
)

# ============================================================
# 19. DATASET HEALTH SCORE
# ============================================================

print("[18/25] Calculating dataset health...")

duplicate_penalty = min(
    exact_duplicates
    /
    len(data),
    1
)

missing_penalty = min(
    missing_rate,
    1
)

anomaly_penalty = (
    data[
        "if_anomaly"
    ].mean()
)

statistical_penalty = (
    data[
        "statistical_outlier"
    ].mean()
)

drift_penalty = min(
    daily_drift_df[
        "average_drift"
    ].mean(),
    1
)

health_score = (
    100
    *
    (
        1
        -
        (
            0.25 * duplicate_penalty
            +
            0.25 * missing_penalty
            +
            0.20 * anomaly_penalty
            +
            0.15 * statistical_penalty
            +
            0.15 * drift_penalty
        )
    )
)

health_score = float(
    np.clip(
        health_score,
        0,
        100
    )
)

# ============================================================
# 20. DATASET DNA
# ============================================================

dataset_dna = {

    "rows":
        int(len(data)),

    "columns":
        int(data.shape[1]),

    "numeric_features":
        int(len(numeric_cols)),

    "categorical_features":
        int(len(categorical_cols)),

    "missing_rate":
        float(missing_rate),

    "exact_duplicate_rate":
        float(
            exact_duplicates
            /
            len(data)
        ),

    "isolation_forest_anomaly_rate":
        float(
            data[
                "if_anomaly"
            ].mean()
        ),

    "statistical_outlier_rate":
        float(
            data[
                "statistical_outlier"
            ].mean()
        ),

    "density_outlier_rate":
        float(
            data[
                "density_outlier"
            ].mean()
        ),

    "average_forensic_score":
        float(
            data[
                "forensic_score"
            ].mean()
        ),

    "critical_records":
        int(
            (
                data[
                    "forensic_level"
                ]
                ==
                "CRITICAL"
            ).sum()
        ),

    "high_risk_records":
        int(
            (
                data[
                    "forensic_level"
                ]
                ==
                "HIGH"
            ).sum()
        ),

    "behavior_clusters":
        int(
            data[
                "behavior_cluster"
            ].nunique()
        ),

    "density_clusters":
        int(
            data[
                "density_cluster"
            ].nunique()
        ),

    "source_count":
        int(
            data[
                "source"
            ].nunique()
        ),

    "dataset_health":
        health_score
}

# ============================================================
# 21. FORENSIC EVIDENCE ENGINE
# ============================================================

print("[19/25] Building evidence engine...")

top_records = (
    data.sort_values(
        "forensic_score",
        ascending=False
    )
    .head(30)
)

evidence_records = []

for _, row in top_records.iterrows():

    evidence = []

    # Isolation Forest
    if row["if_anomaly"]:

        evidence.append(
            "Isolation Forest classified "
            "the record as anomalous"
        )

    # Robust statistical deviation
    if row["max_robust_z"] >= 3.5:

        evidence.append(
            f"Robust statistical deviation "
            f"(max z={row['max_robust_z']:.2f})"
        )

    # Missingness
    row_missing = (
        data.loc[
            data["record_id"]
            ==
            row["record_id"],
            numeric_cols
        ]
        .isna()
        .sum()
        .sum()
    )

    if row_missing > 0:

        evidence.append(
            f"{row_missing} missing numeric values"
        )

    # Density
    if row["density_outlier"]:

        evidence.append(
            "Record lies outside dense "
            "behavioral regions"
        )

    # Rare source
    source_frequency = (
        data[
            "source"
        ]
        .value_counts(
            normalize=True
        )
        .get(
            row["source"],
            0
        )
    )

    if source_frequency < 0.05:

        evidence.append(
            "Record belongs to a relatively "
            "rare source group"
        )

    if not evidence:

        evidence.append(
            "No single dominant forensic signal"
        )

    evidence_records.append({

        "record_id":
            int(row["record_id"]),

        "timestamp":
            str(row["timestamp"]),

        "source":
            row["source"],

        "forensic_score":
            float(
                row["forensic_score"]
            ),

        "classification":
            row["forensic_level"],

        "evidence":
            " | ".join(evidence)
    })

evidence_df = pd.DataFrame(
    evidence_records
)

# ============================================================
# 22. INVESTIGATION CASES
# ============================================================

print("[20/25] Generating investigation cases...")

cases = []

# ------------------------------------------------------------
# Case 1: Missingness
# ------------------------------------------------------------

if missing_rate > 0:

    top_missing = (
        missing_df
        .iloc[0]
    )

    cases.append({

        "case_id":
            "ORIGIN-CASE-001",

        "case_type":
            "DATA_QUALITY",

        "severity":
            "MEDIUM",

        "title":
            "Missingness Pattern Detected",

        "description":
            (
                f"Feature "
                f"{top_missing['feature']} "
                f"contains "
                f"{top_missing['missing_rate']:.2%} "
                f"missing values."
            ),

        "status":
            "OPEN"
    })

# ------------------------------------------------------------
# Case 2: Anomaly concentration
# ------------------------------------------------------------

if data["if_anomaly"].mean() > 0.02:

    cases.append({

        "case_id":
            "ORIGIN-CASE-002",

        "case_type":
            "ANOMALY",

        "severity":
            "HIGH",

        "title":
            "Elevated Anomaly Concentration",

        "description":
            (
                f"Isolation Forest identified "
                f"{data['if_anomaly'].sum():,} "
                f"potential anomalous records."
            ),

        "status":
            "OPEN"
    })

# ------------------------------------------------------------
# Case 3: Distribution shift
# ------------------------------------------------------------

shifted_features = (
    distribution_shift_df[
        distribution_shift_df[
            "shift_detected"
        ]
    ]
)

if len(shifted_features) > 0:

    cases.append({

        "case_id":
            "ORIGIN-CASE-003",

        "case_type":
            "DISTRIBUTION_SHIFT",

        "severity":
            "HIGH",

        "title":
            "Distribution Shift Detected",

        "description":
            (
                f"{len(shifted_features)} "
                f"features show statistically "
                f"significant distribution differences "
                f"between baseline and later observations."
            ),

        "status":
            "INVESTIGATE"
    })

# ------------------------------------------------------------
# Case 4: Duplicate pattern
# ------------------------------------------------------------

if exact_duplicates > 0:

    cases.append({

        "case_id":
            "ORIGIN-CASE-004",

        "case_type":
            "DUPLICATES",

        "severity":
            "MEDIUM",

        "title":
            "Duplicate Records Detected",

        "description":
            (
                f"{exact_duplicates:,} exact duplicate "
                f"records were found."
            ),

        "status":
            "OPEN"
    })

# ------------------------------------------------------------
# Case 5: Strong dependencies
# ------------------------------------------------------------

if len(dependency_df) > 0:

    strongest = (
        dependency_df
        .sort_values(
            "correlation",
            key=lambda x: abs(x),
            ascending=False
        )
        .iloc[0]
    )

    cases.append({

        "case_id":
            "ORIGIN-CASE-005",

        "case_type":
            "FEATURE_DEPENDENCY",

        "severity":
            "LOW",

        "title":
            "Strong Feature Dependency",

        "description":
            (
                f"{strongest['feature_1']} and "
                f"{strongest['feature_2']} have "
                f"correlation "
                f"{strongest['correlation']:.3f}."
            ),

        "status":
            "REVIEW"
    })

cases_df = pd.DataFrame(
    cases
)

# ============================================================
# 23. EXECUTIVE METRICS
# ============================================================

print("[21/25] Preparing executive metrics...")

critical_count = int(
    (
        data[
            "forensic_level"
        ]
        ==
        "CRITICAL"
    ).sum()
)

high_count = int(
    (
        data[
            "forensic_level"
        ]
        ==
        "HIGH"
    ).sum()
)

medium_count = int(
    (
        data[
            "forensic_level"
        ]
        ==
        "MEDIUM"
    ).sum()
)

low_count = int(
    (
        data[
            "forensic_level"
        ]
        ==
        "LOW"
    ).sum()
)

# ============================================================
# 24. VISUALIZATION DASHBOARD
# ============================================================

print("[22/25] Building visual intelligence dashboard...")

# ------------------------------------------------------------
# Dashboard 1 — Dataset Command Center
# ------------------------------------------------------------

fig = plt.figure(
    figsize=(18, 10)
)

fig.suptitle(
    "ORIGIN AI — DATASET FORENSIC COMMAND CENTER",
    fontsize=22,
    fontweight="bold"
)

dashboard_metrics = [

    (
        "Dataset Health",
        f"{health_score:.1f}/100"
    ),

    (
        "Records",
        f"{len(data):,}"
    ),

    (
        "Anomalies",
        f"{data['if_anomaly'].sum():,}"
    ),

    (
        "Exact Duplicates",
        f"{exact_duplicates:,}"
    ),

    (
        "Missing Rate",
        f"{missing_rate:.2%}"
    ),

    (
        "Critical",
        f"{critical_count:,}"
    ),

    (
        "High Risk",
        f"{high_count:,}"
    ),

    (
        "Clusters",
        f"{data['behavior_cluster'].nunique()}"
    )
]

for i, (
    label,
    value
) in enumerate(
    dashboard_metrics
):

    ax = fig.add_subplot(
        2,
        4,
        i + 1
    )

    ax.text(
        0.5,
        0.60,
        value,
        ha="center",
        va="center",
        fontsize=25,
        fontweight="bold"
    )

    ax.text(
        0.5,
        0.27,
        label,
        ha="center",
        va="center",
        fontsize=12
    )

    ax.set_xticks([])
    ax.set_yticks([])

    for spine in ax.spines.values():

        spine.set_visible(True)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Visualization 2 — PCA Crime Scene
# ------------------------------------------------------------

plt.figure(
    figsize=(13, 9)
)

normal_points = data[
    ~data["if_anomaly"]
]

anomaly_points = data[
    data["if_anomaly"]
]

plt.scatter(
    normal_points["PCA_1"],
    normal_points["PCA_2"],
    s=12,
    alpha=0.30,
    label="Normal"
)

plt.scatter(
    anomaly_points["PCA_1"],
    anomaly_points["PCA_2"],
    s=55,
    marker="x",
    label="Isolation Forest Anomaly"
)

plt.title(
    "ORIGIN AI — Dataset Crime Scene",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel(
    f"Principal Component 1 "
    f"({pc1_variance:.1%})"
)

plt.ylabel(
    f"Principal Component 2 "
    f"({pc2_variance:.1%})"
)

plt.legend()
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Visualization 3 — Forensic Score Distribution
# ------------------------------------------------------------

plt.figure(
    figsize=(13, 6)
)

plt.hist(
    data["forensic_score"],
    bins=50,
    alpha=0.80
)

plt.axvline(
    0.35,
    linestyle="--",
    label="Medium"
)

plt.axvline(
    0.60,
    linestyle="--",
    label="High"
)

plt.axvline(
    0.80,
    linestyle=":",
    label="Critical"
)

plt.title(
    "ORIGIN AI — Record Forensic Score Distribution",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel(
    "Forensic Score"
)

plt.ylabel(
    "Number of Records"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Visualization 4 — Temporal Drift
# ------------------------------------------------------------

plt.figure(
    figsize=(16, 6)
)

plt.plot(
    pd.to_datetime(
        daily_drift_df["date"]
    ),
    daily_drift_df[
        "average_drift"
    ],
    linewidth=2,
    label="Average Drift"
)

plt.plot(
    pd.to_datetime(
        daily_drift_df["date"]
    ),
    daily_drift_df[
        "maximum_drift"
    ],
    linewidth=1.5,
    label="Maximum Drift"
)

plt.title(
    "ORIGIN AI — Dataset Evolution & Temporal Drift",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel(
    "Date"
)

plt.ylabel(
    "Relative Distribution Change"
)

plt.legend()

plt.grid(alpha=0.25)

plt.xticks(
    rotation=30
)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Visualization 5 — Feature Correlation
# ------------------------------------------------------------

plt.figure(
    figsize=(13, 11)
)

plt.imshow(
    correlation,
    aspect="auto"
)

plt.colorbar(
    label="Correlation"
)

plt.xticks(
    range(
        len(numeric_cols)
    ),
    numeric_cols,
    rotation=75,
    ha="right"
)

plt.yticks(
    range(
        len(numeric_cols)
    ),
    numeric_cols
)

plt.title(
    "ORIGIN AI — Feature Dependency Map",
    fontsize=17,
    fontweight="bold"
)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Visualization 6 — Behavioral Clusters
# ------------------------------------------------------------

plt.figure(
    figsize=(13, 9)
)

for cluster in sorted(
    data[
        "behavior_cluster"
    ].unique()
):

    subset = data[
        data[
            "behavior_cluster"
        ]
        ==
        cluster
    ]

    plt.scatter(
        subset["PCA_1"],
        subset["PCA_2"],
        s=14,
        alpha=0.40,
        label=f"Cluster {cluster}"
    )

plt.title(
    "ORIGIN AI — Behavioral Dataset Fingerprints",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel("PC1")
plt.ylabel("PC2")

plt.legend()

plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Visualization 7 — Source Fingerprints
# ------------------------------------------------------------

fingerprint_features = [
    "age",
    "activity_score",
    "risk_score",
    "transaction_count",
    "feature_01",
    "feature_03",
    "feature_07"
]

fingerprint_profile = (
    data.groupby(
        "source"
    )[
        fingerprint_features
    ]
    .mean()
)

fingerprint_scaled = pd.DataFrame(
    StandardScaler().fit_transform(
        fingerprint_profile
    ),
    index=fingerprint_profile.index,
    columns=fingerprint_profile.columns
)

plt.figure(
    figsize=(15, 7)
)

for source in (
    fingerprint_scaled.index
):

    plt.plot(
        fingerprint_features,
        fingerprint_scaled.loc[
            source
        ],
        marker="o",
        linewidth=2,
        label=source
    )

plt.axhline(
    0,
    linestyle="--"
)

plt.title(
    "ORIGIN AI — Source Statistical Fingerprints",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel(
    "Feature"
)

plt.ylabel(
    "Standardized Mean"
)

plt.legend()

plt.xticks(
    rotation=30
)

plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Visualization 8 — Missingness
# ------------------------------------------------------------

missing_positive = (
    missing_df[
        missing_df[
            "missing_count"
        ] > 0
    ]
)

plt.figure(
    figsize=(13, 6)
)

plt.bar(
    missing_positive[
        "feature"
    ],
    missing_positive[
        "missing_count"
    ]
)

plt.title(
    "ORIGIN AI — Missing Value Forensics",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel(
    "Feature"
)

plt.ylabel(
    "Missing Records"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Visualization 9 — Risk Distribution
# ------------------------------------------------------------

risk_counts = pd.Series({
    "LOW": low_count,
    "MEDIUM": medium_count,
    "HIGH": high_count,
    "CRITICAL": critical_count
})

plt.figure(
    figsize=(10, 6)
)

plt.bar(
    risk_counts.index,
    risk_counts.values
)

plt.title(
    "ORIGIN AI — Forensic Risk Classification",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel(
    "Classification"
)

plt.ylabel(
    "Records"
)

plt.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Visualization 10 — Top suspicious records
# ------------------------------------------------------------

top_15 = (
    data.sort_values(
        "forensic_score",
        ascending=False
    )
    .head(15)
)

plt.figure(
    figsize=(14, 7)
)

plt.bar(
    top_15[
        "record_id"
    ].astype(str),
    top_15[
        "forensic_score"
    ]
)

plt.title(
    "ORIGIN AI — Highest Priority Records",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel(
    "Record ID"
)

plt.ylabel(
    "Forensic Score"
)

plt.xticks(
    rotation=45
)

plt.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Visualization 11 — Distribution comparison
# ------------------------------------------------------------

selected_distribution_features = [
    "age",
    "activity_score",
    "risk_score",
    "transaction_count",
    "feature_01",
    "feature_07"
]

fig, axes = plt.subplots(
    2,
    3,
    figsize=(17, 10)
)

axes = axes.flatten()

baseline_mask = (
    data["timestamp"]
    <
    pd.Timestamp("2026-03-01")
)

later_mask = ~baseline_mask

for ax, feature in zip(
    axes,
    selected_distribution_features
):

    ax.hist(
        data.loc[
            baseline_mask,
            feature
        ].dropna(),
        bins=30,
        alpha=0.50,
        label="Baseline"
    )

    ax.hist(
        data.loc[
            later_mask,
            feature
        ].dropna(),
        bins=30,
        alpha=0.50,
        label="Later"
    )

    ax.set_title(
        feature
    )

    ax.legend()

    ax.grid(
        axis="y",
        alpha=0.20
    )

plt.suptitle(
    "ORIGIN AI — Distribution Shift Investigation",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Visualization 12 — Missingness pattern
# ------------------------------------------------------------

missing_matrix = (
    data[
        numeric_cols
    ]
    .isna()
    .astype(int)
    .head(400)
)

plt.figure(
    figsize=(15, 7)
)

plt.imshow(
    missing_matrix.T,
    aspect="auto"
)

plt.colorbar(
    label="Missing Indicator"
)

plt.yticks(
    range(
        len(numeric_cols)
    ),
    numeric_cols
)

plt.xlabel(
    "Records"
)

plt.ylabel(
    "Features"
)

plt.title(
    "ORIGIN AI — Missingness Pattern Forensics",
    fontsize=17,
    fontweight="bold"
)

plt.tight_layout()
plt.show()

# ============================================================
# 25. FINAL TABLES + REPORT
# ============================================================

print("[23/25] Preparing final forensic tables...")

# ------------------------------------------------------------
# Dataset DNA table
# ------------------------------------------------------------

dna_df = pd.DataFrame(
    [
        {
            "metric":
                key,

            "value":
                value
        }

        for key, value
        in dataset_dna.items()
    ]
)

# ------------------------------------------------------------
# Investigation summary
# ------------------------------------------------------------

investigation_summary = pd.DataFrame({

    "Metric": [

        "Dataset Health",

        "Total Records",

        "Total Columns",

        "Missing Rate",

        "Exact Duplicate Rate",

        "Isolation Forest Anomaly Rate",

        "Statistical Outlier Rate",

        "Density Outlier Rate",

        "Critical Records",

        "High-Risk Records",

        "Behavior Clusters",

        "Distribution-Shifting Features"

    ],

    "Value": [

        f"{health_score:.2f}/100",

        f"{len(data):,}",

        f"{data.shape[1]}",

        f"{missing_rate:.2%}",

        f"{exact_duplicates / len(data):.2%}",

        f"{data['if_anomaly'].mean():.2%}",

        f"{data['statistical_outlier'].mean():.2%}",

        f"{data['density_outlier'].mean():.2%}",

        f"{critical_count:,}",

        f"{high_count:,}",

        f"{data['behavior_cluster'].nunique()}",

        f"{len(shifted_features):,}"
    ]
})

# ------------------------------------------------------------
# Print important results
# ------------------------------------------------------------

print("\n")
print("=" * 110)
print("ORIGIN AI — DATASET FORENSIC SUMMARY")
print("=" * 110)

display(
    investigation_summary
)

print("\n")
print("=" * 110)
print("TOP SUSPICIOUS RECORDS")
print("=" * 110)

display(
    evidence_df.head(15)
)

print("\n")
print("=" * 110)
print("INVESTIGATION CASES")
print("=" * 110)

if len(cases_df) > 0:
    display(
        cases_df
    )
else:
    print(
        "No investigation cases generated."
    )

# ============================================================
# AUTOMATED FORENSIC REPORT
# ============================================================

print("[24/25] Writing forensic report...")

top_record = (
    data.sort_values(
        "forensic_score",
        ascending=False
    )
    .iloc[0]
)

top_missing_feature = (
    missing_df.iloc[0]
    if len(missing_df)
    else None
)

if (
    len(dependency_df)
    >
    0
):

    strongest_dependency = (
        dependency_df
        .sort_values(
            "correlation",
            key=lambda x: abs(x),
            ascending=False
        )
        .iloc[0]
    )

    dependency_text = (
        f"{strongest_dependency['feature_1']} "
        f"and "
        f"{strongest_dependency['feature_2']} "
        f"(correlation="
        f"{strongest_dependency['correlation']:.3f})"
    )

else:

    dependency_text = (
        "No strong dependency detected."
    )

report = f"""
============================================================
ORIGIN AI
DATASET FORENSIC INVESTIGATION REPORT
============================================================

PROJECT
-------
{PROJECT_NAME}

Version:
{PROJECT_VERSION}

Analysis Mode:
Synthetic Demonstration

============================================================
1. DATASET DNA
============================================================

Rows:
{len(data):,}

Columns:
{data.shape[1]}

Numeric Features:
{len(numeric_cols)}

Categorical Features:
{len(categorical_cols)}

Sources:
{data['source'].nunique()}

Behavioral Clusters:
{data['behavior_cluster'].nunique()}

============================================================
2. DATA QUALITY
============================================================

Missing Cell Rate:
{missing_rate:.2%}

Exact Duplicate Count:
{exact_duplicates:,}

Exact Duplicate Rate:
{exact_duplicates / len(data):.2%}

Isolation Forest Anomaly Rate:
{data['if_anomaly'].mean():.2%}

Statistical Outlier Rate:
{data['statistical_outlier'].mean():.2%}

Density Outlier Rate:
{data['density_outlier'].mean():.2%}

============================================================
3. DATASET HEALTH
============================================================

Overall Dataset Health:
{health_score:.2f}/100

============================================================
4. FORENSIC INTELLIGENCE
============================================================

Critical Records:
{critical_count:,}

High-Risk Records:
{high_count:,}

Medium-Risk Records:
{medium_count:,}

Low-Risk Records:
{low_count:,}

Average Forensic Score:
{data['forensic_score'].mean():.4f}

Maximum Forensic Score:
{data['forensic_score'].max():.4f}

============================================================
5. TOP INVESTIGATION TARGET
============================================================

Record ID:
{top_record['record_id']}

Timestamp:
{top_record['timestamp']}

Source:
{top_record['source']}

Forensic Score:
{top_record['forensic_score']:.4f}

Classification:
{top_record['forensic_level']}

Maximum Robust Z-Score:
{top_record['max_robust_z']:.3f}

Isolation Forest Anomaly:
{top_record['if_anomaly']}

Behavior Cluster:
{top_record['behavior_cluster']}

============================================================
6. MISSINGNESS FINDING
============================================================

Most affected feature:
{top_missing_feature['feature'] if top_missing_feature is not None else 'None'}

Missing records:
{int(top_missing_feature['missing_count']) if top_missing_feature is not None else 0}

============================================================
7. FEATURE DEPENDENCY
============================================================

Strongest detected dependency:

{dependency_text}

============================================================
8. DISTRIBUTION SHIFT
============================================================

Features with statistically significant
baseline/later distribution differences:

{len(shifted_features)}

============================================================
9. METHODS USED
============================================================

- Dataset profiling
- Missing-value analysis
- Exact duplicate analysis
- Robust Z-score analysis
- Isolation Forest
- PCA
- KMeans clustering
- DBSCAN density analysis
- Correlation analysis
- Temporal drift analysis
- Kolmogorov-Smirnov distribution testing
- Source fingerprinting
- Composite forensic scoring

============================================================
10. INTERPRETATION
============================================================

ORIGIN AI combines multiple statistical and machine-learning
signals to identify records, feature distributions, and
dataset regions that deserve further investigation.

The system does not automatically determine:

- fraud
- malicious activity
- data provenance
- intentional manipulation
- causal relationships

An anomaly is an analytical signal, not a final conclusion.

Human/domain review is required before making consequential
decisions.

============================================================
END OF REPORT
============================================================
"""

report_path = os.path.join(
    EXPORT_DIR,
    "origin_forensic_report.txt"
)

with open(
    report_path,
    "w",
    encoding="utf-8"
) as f:

    f.write(report)

# ============================================================
# 26. EXPORT ALL RESULTS
# ============================================================

print("[25/25] Exporting analytical artifacts...")

data.to_csv(
    os.path.join(
        EXPORT_DIR,
        "origin_full_forensic_dataset.csv"
    ),
    index=False
)

profile_df.to_csv(
    os.path.join(
        EXPORT_DIR,
        "origin_column_profile.csv"
    ),
    index=False
)

missing_df.to_csv(
    os.path.join(
        EXPORT_DIR,
        "origin_missing_values.csv"
    ),
    index=False
)

duplicate_rows_df.to_csv(
    os.path.join(
        EXPORT_DIR,
        "origin_duplicate_records.csv"
    ),
    index=False
)

dependency_df.to_csv(
    os.path.join(
        EXPORT_DIR,
        "origin_feature_dependencies.csv"
    ),
    index=False
)

distribution_shift_df.to_csv(
    os.path.join(
        EXPORT_DIR,
        "origin_distribution_shift.csv"
    ),
    index=False
)

daily_drift_df.to_csv(
    os.path.join(
        EXPORT_DIR,
        "origin_temporal_drift.csv"
    ),
    index=False
)

source_profile.to_csv(
    os.path.join(
        EXPORT_DIR,
        "origin_source_fingerprints.csv"
    )
)

source_distance_matrix.to_csv(
    os.path.join(
        EXPORT_DIR,
        "origin_source_distance_matrix.csv"
    )
)

evidence_df.to_csv(
    os.path.join(
        EXPORT_DIR,
        "origin_forensic_evidence.csv"
    ),
    index=False
)

cases_df.to_csv(
    os.path.join(
        EXPORT_DIR,
        "origin_investigation_cases.csv"
    ),
    index=False
)

dna_df.to_csv(
    os.path.join(
        EXPORT_DIR,
        "origin_dataset_dna.csv"
    ),
    index=False
)

investigation_summary.to_csv(
    os.path.join(
        EXPORT_DIR,
        "origin_executive_summary.csv"
    ),
    index=False
)

with open(
    os.path.join(
        EXPORT_DIR,
        "origin_dataset_dna.json"
    ),
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        dataset_dna,
        f,
        indent=4
    )

# ============================================================
# FINAL COMMAND CENTER
# ============================================================

print("\n")
print("=" * 110)
print("ORIGIN AI — FINAL FORENSIC COMMAND CENTER")
print("=" * 110)

print(f"""
PROJECT
-------
{PROJECT_NAME}
Version {PROJECT_VERSION}

DATASET
-------
Records                 : {len(data):,}
Columns                 : {data.shape[1]}
Numeric Features        : {len(numeric_cols)}
Categorical Features    : {len(categorical_cols)}
Sources                 : {data['source'].nunique()}

QUALITY
-------
Dataset Health          : {health_score:.2f}/100
Missing Rate            : {missing_rate:.2%}
Exact Duplicates        : {exact_duplicates:,}

ANOMALY INTELLIGENCE
--------------------
Isolation Forest        : {data['if_anomaly'].sum():,}
Statistical Outliers    : {data['statistical_outlier'].sum():,}
Density Outliers        : {data['density_outlier'].sum():,}

FORENSIC CLASSIFICATION
-----------------------
Critical                : {critical_count:,}
High                    : {high_count:,}
Medium                  : {medium_count:,}
Low                     : {low_count:,}

DISCOVERED STRUCTURE
--------------------
Behavior Clusters       : {data['behavior_cluster'].nunique()}
Density Clusters        : {data['density_cluster'].nunique()}
Strong Dependencies     : {len(dependency_df)}
Distribution Shifts     : {len(shifted_features)}

TOP INVESTIGATION TARGET
------------------------
Record ID               : {top_record['record_id']}
Source                  : {top_record['source']}
Timestamp               : {top_record['timestamp']}
Forensic Score          : {top_record['forensic_score']:.4f}
Classification          : {top_record['forensic_level']}

OUTPUT DIRECTORY
----------------
{EXPORT_DIR}/

Generated:
- origin_full_forensic_dataset.csv
- origin_column_profile.csv
- origin_missing_values.csv
- origin_duplicate_records.csv
- origin_feature_dependencies.csv
- origin_distribution_shift.csv
- origin_temporal_drift.csv
- origin_source_fingerprints.csv
- origin_source_distance_matrix.csv
- origin_forensic_evidence.csv
- origin_investigation_cases.csv
- origin_dataset_dna.csv
- origin_dataset_dna.json
- origin_executive_summary.csv
- origin_forensic_report.txt

============================================================
ORIGIN AI ANALYSIS COMPLETE
============================================================

Every dataset has a history.
ORIGIN reconstructs the evidence.

IMPORTANT:
All findings are analytical signals generated from synthetic
data and should be interpreted as investigation leads rather
than definitive conclusions.
============================================================
""")
```
